```python
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, clear_output
import json
import ast

if "df_tmdb" not in globals():
    df_tmdb = pd.read_csv("/Users/roop/Library/CloudStorage/OneDrive-UniversityofFlorida/Courses/Spring 26 EGN6933/Code/CineMatch/Data/outputs/tmdb_semantic_catalog_alllangs_with_new_movies.csv")

search_input = widgets.Text(
    description="Title:",
    placeholder="Enter movie title"
)
search_button = widgets.Button(
    description="Search",
    button_style="primary"
)
matches_select = widgets.Select(
    description="Matches:",
    options=[],
    rows=12,
    layout=widgets.Layout(width="98%")
)

search_output = widgets.Output()
details_output = widgets.Output()

def parse_json_like(value):
    if pd.isna(value):
        return None
    if isinstance(value, (list, dict)):
        return value
    if isinstance(value, str):
        text = value.strip()
        if not text:
            return None
        for parser in (json.loads, ast.literal_eval):
            try:
                return parser(text)
            except Exception:
                pass
    return value

def format_list_or_dict(value):
    parsed = parse_json_like(value)

    if parsed is None:
        return "N/A"

    if isinstance(parsed, list):
        items = []
        for item in parsed:
            if isinstance(item, dict):
                if item.get("name"):
                    items.append(str(item["name"]))
                elif item.get("english_name"):
                    items.append(str(item["english_name"]))
                else:
                    items.append(str(item))
            else:
                items.append(str(item))
        return ", ".join(items) if items else "[]"

    if isinstance(parsed, dict):
        return ", ".join([f"{k}: {v}" for k, v in parsed.items()])

    return str(parsed)

def format_value(column, value):
    if pd.isna(value):
        return "N/A"

    if column in ["budget", "revenue"]:
        try:
            num = float(value)
            return "${:,.0f}".format(num)
        except Exception:
            return str(value)

    if column in [
        "genres",
        "production_companies",
        "production_countries",
        "spoken_languages",
        "keywords",
    ]:
        return format_list_or_dict(value)

    return str(value)

def show_movie_details(row_index):
    with details_output:
        clear_output()
        row = df_tmdb.loc[row_index]

        details = pd.DataFrame({
            "Feature": df_tmdb.columns,
            "Value": [format_value(col, row[col]) for col in df_tmdb.columns]
        })
        display(details)

def on_search_click(_):
    with search_output:
        clear_output()
        details_output.clear_output()

        query = search_input.value.strip()
        if not query:
            print("Please enter a title.")
            matches_select.options = []
            return

        mask = df_tmdb["title"].astype(str).str.contains(query, case=False, na=False)
        found = df_tmdb[mask].copy()

        if found.empty:
            print("No matches found.")
            matches_select.options = []
            return

        exact_mask = found["title"].astype(str).str.lower().eq(query.lower())
        found = pd.concat([found[exact_mask], found[~exact_mask]], axis=0).head(200)

        options = []
        for idx, movie in found.iterrows():
            release_date = movie.get("release_date", "")
            year = str(release_date)[:4] if pd.notna(release_date) and str(release_date) else "----"
            label = f"{movie.get('title', 'Unknown')} ({year}) | id={movie.get('id', 'N/A')}"
            options.append((label, idx))

        matches_select.options = options
        matches_select.value = options[0][1]
        print(f"Found {len(found)} match(es). Select one movie to see all features.")

def on_movie_selected(change):
    if change["name"] == "value" and change["new"] is not None:
        show_movie_details(change["new"])

search_button.on_click(on_search_click)
matches_select.observe(on_movie_selected, names="value")

display(
    widgets.VBox([
        widgets.HBox([search_input, search_button]),
        search_output,
        matches_select,
        details_output
    ])
)
```
