```python
from huggingface_hub import create_repo

repo_id = "ml8r/cinematch" # e.g., "johndoe/my-25gb-data"

# Create the repository (set private=True if you don't want it public)
create_repo(repo_id, repo_type="dataset", private=False, exist_ok=True)
print(f"Repository {repo_id} is ready!")
```

```python
from huggingface_hub import HfApi

api = HfApi()

# This points to the parent folder containing your two subfolders
drive_folder_path = "/content/drive/MyDrive/cinematch/"
repo_id = "ml8r/cinematch"

print("Starting upload... This may take a little while.")

# Upload the contents of cinematch/ directly to the repo root
api.upload_folder(
    folder_path=drive_folder_path,
    repo_id=repo_id,
    repo_type="dataset",
    path_in_repo="."  # Using "." keeps your internal folder structure intact
)

print("Upload complete! Your folders are now in the root of the repository.")
```
